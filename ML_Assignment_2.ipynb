```python

import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Scikit-learn:", sklearn.__version__)
print("All required basic libraries imported successfully!")
```

```python

from sklearn.datasets import load_breast_cancer
import pandas as pd

# Load the dataset
data = load_breast_cancer()

# Convert the data into a table
df = pd.DataFrame(data.data, columns=data.feature_names)

# Add the target column
df["target"] = data.target

# Display the first five rows
df.head()
```

```python

print("Dataset shape:", df.shape)
print("\nNumber of rows:", df.shape[0])
print("Number of input features:", df.shape[1] - 1)

print("\nTarget counts:")
print(df["target"].value_counts())
```

```python

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python

X = df.drop("target", axis=1)
y = df["target"]

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python

from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Learn scaling values from training data only
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling to test data
X_test_scaled = scaler.transform(X_test)

print("Scaled training data:", X_train_scaled.shape)
print("Scaled testing data:", X_test_scaled.shape)
```

```python

from sklearn.linear_model import LogisticRegression

# Create the model
lr_model = LogisticRegression(max_iter=2000, random_state=42)

# Train the model
lr_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# Predict the test-set classes
y_pred_lr = lr_model.predict(X_test_scaled)

# Predict probabilities for AUC
y_prob_lr = lr_model.predict_proba(X_test_scaled)[:, 1]

print("Logistic Regression Results")
print("---------------------------")
print("Accuracy:", round(accuracy_score(y_test, y_pred_lr), 4))
print("AUC:", round(roc_auc_score(y_test, y_prob_lr), 4))
print("Precision:", round(precision_score(y_test, y_pred_lr), 4))
print("Recall:", round(recall_score(y_test, y_pred_lr), 4))
print("F1 Score:", round(f1_score(y_test, y_pred_lr), 4))
print("MCC:", round(matthews_corrcoef(y_test, y_pred_lr), 4))
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

# 1. Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# 2. K-Nearest Neighbors (KNN)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled, y_train)

# 3. Gaussian Naive Bayes
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# 4. Support Vector Machine (SVM)
svm_model = SVC(probability=True, random_state=42)
svm_model.fit(X_train_scaled, y_train)

print("All four additional models trained successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score, roc_auc_score, precision_score,
    recall_score, f1_score, matthews_corrcoef
)
import pandas as pd

models = {
    "Logistic Regression": (lr_model, X_test_scaled),
    "Decision Tree": (dt_model, X_test),
    "KNN": (knn_model, X_test_scaled),
    "Gaussian Naive Bayes": (nb_model, X_test),
    "SVM": (svm_model, X_test_scaled)
}

results = []

for name, (model, test_features) in models.items():
    predictions = model.predict(test_features)
    probabilities = model.predict_proba(test_features)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "AUC": roc_auc_score(y_test, probabilities),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "F1 Score": f1_score(y_test, predictions, zero_division=0),
        "MCC": matthews_corrcoef(y_test, predictions)
    })

results_df = pd.DataFrame(results)
print(results_df.round(4).to_string(index=False))
```

```python

from sklearn.metrics import confusion_matrix, classification_report
import pandas as pd

# Check Logistic Regression predictions
cm = confusion_matrix(y_test, y_pred_lr, labels=[0, 1])

print("Confusion Matrix (rows = actual, columns = predicted)")
print("Class order: 0 = malignant, 1 = benign")
print(cm)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_lr,
    target_names=["Malignant", "Benign"]
))
```

```python

# Save the comparison table as a CSV file
results_df.to_csv("model_comparison.csv", index=False)

print("Saved model_comparison.csv successfully!")
```

```python

import joblib
import os

# Create a folder to store the trained models
os.makedirs("saved_models", exist_ok=True)

# Save each trained model
joblib.dump(lr_model, "saved_models/logistic_regression.pkl")
joblib.dump(dt_model, "saved_models/decision_tree.pkl")
joblib.dump(knn_model, "saved_models/knn.pkl")
joblib.dump(nb_model, "saved_models/naive_bayes.pkl")
joblib.dump(svm_model, "saved_models/svm.pkl")

# Save the scaler used for Logistic Regression, KNN and SVM
joblib.dump(scaler, "saved_models/scaler.pkl")

print("All five models and the scaler have been saved!")
```
